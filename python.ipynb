```python
#basic function
def greet():
    print("Hello")
    print("Welcome to python")
greet()
```

```python
#Funtion without parameters
def welcome():
    print("Welcome to python")

welcome()
welcome()
welcome()
```

```python
#Funtions with parameters
def greet(name):
    print("Hello", name)

greet("haasini")
greet("sravyaa")
```

```python
#Multiple parameters
def add(a, b):
    print("Sum:", a + b)


add(10,20)
add(50,30)
```

```python
#Funtions with return value
def add(a, b):
    return a + b

result = add(10, 20)

print(result)
```

```python
def add(a, b):
    return a + b

result = add(10, 20)

if result > 25:
    print("Large result")
```

```python
#Funtion for Even/Odd
def check_even_odd(number):

    if number % 2 == 0:
         return "Even"
    else:
         return "Odd"

result = check_even_odd(25)

print(result)
```

```python
# Function for Pass or Fail

def check_pass_fail(marks):
    if marks >= 40:
        print("Pass")
    else:
        print("Fail")

check_pass_fail(75)
check_pass_fail(35)
```

```python
#funtions for Largest of two numbers
def largest(a, b):

    if a > b:
        return a
    else:
        return b

result = largest(50, 30)

print("Largest:" , result)
```
