```python
#dictionaries in python
#Dictionary is a collection of key value pairs that is unordered and mutable
student = {
    "name": "haasini",
    "age" : 18,
    "course" : "python"

}
print(student)
```

```python
#access element in dic
print(student["name"])
print(student["age"])
print(student["course"])
```

```python
#add new data to dictionary
student["city"] = "vijayawada" 

print(student)
```

```python
student = {
     "name" : "haasini",
      "age": 18,
      "course" : "python"
}

student.popitem()

print(student)
```

```python
student = {
     "name" : "haasini",
      "age": 18,
      "course" : "python"
}
print(student.keys())
#keys() returns all the keys in the dictionary

print(student.values())
#items() returns all the values in the dictionary

print(student.items())
#items() returns all key-value pairs

print(student.get("name"))
#get()returns the value of the specified key

student.update({"age": 18})
#update() update the value of the specified key

print(student)

student.pop("age")
#pop() removes the specified key and its value

print(student)
```

```python
student = {
     "name": "haasini"
}

student.setdefault("age", 18)

print(student)
```
